# 04 - Patient Phenotype Clustering & Biometric Anomaly Detection
Unsupervised patient segmentation using K-Means, Gaussian Mixture Models (GMM), and HDBSCAN, evaluating optimal $k$ with Silhouette score and Davies-Bouldin index, 2D PCA projection, semantic auto-labeling, and anomaly detection via Isolation Forest.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
from src.ml.datasets import generate_synthetic_clinical_cohort
from src.ml.preprocessor import build_preprocessor_pipeline, NUMERIC_FEATURES, CATEGORICAL_FEATURES
from src.ml.clustering import evaluate_optimal_k, PatientPhenotypeSegmenter
```

### 1. Optimal k Selection (Silhouette & Davies-Bouldin)


```python
df = generate_synthetic_clinical_cohort(n_samples=1500, random_state=42)
preprocessor = build_preprocessor_pipeline()
X = preprocessor.fit_transform(df[NUMERIC_FEATURES + CATEGORICAL_FEATURES])
best_k, sil_scores, db_scores = evaluate_optimal_k(X, k_range=range(2, 6))
print(f'Optimal k: {best_k}')
print('Silhouette Scores:', sil_scores)
print('Davies-Bouldin Scores:', db_scores)
```

### 2. Auto-Labeled Phenotype Profiles


```python
feature_names = list(preprocessor.get_feature_names_out())
segmenter = PatientPhenotypeSegmenter(n_clusters=best_k, random_state=42)
segmenter.fit(X, feature_names)
for cid, prof in segmenter.cluster_profiles.items():
    print(f'Cluster {cid}: {prof["label"]} (Size={prof["size"]})')
```
