# 01 - SwasthAI Clinical Data Exploration & Preprocessing
This notebook explores the clinical biometrics and epidemiological distributions (ICMR-INDIAB, Pima Indian Diabetes, Framingham, and UCI Heart Disease), handles missing values using **IterativeImputer (MICE)**, and addresses class imbalance with **SMOTE**.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.ml.datasets import generate_synthetic_clinical_cohort, save_dataset_manifest
from src.ml.preprocessor import build_preprocessor_pipeline, balance_training_data, NUMERIC_FEATURES, CATEGORICAL_FEATURES
```

### 1. Generate & Inspect Indian Cardiometabolic Cohort


```python
df = generate_synthetic_clinical_cohort(n_samples=2500, random_state=42)
print(f'Cohort Dimensions: {df.shape}')
df.head()
```

### 2. Missing Value Analysis & Iterative Imputer (MICE)


```python
print('Missing Values Count:\n', df[NUMERIC_FEATURES].isnull().sum())
preprocessor = build_preprocessor_pipeline(use_iterative_imputer=True)
X_trans = preprocessor.fit_transform(df[NUMERIC_FEATURES + CATEGORICAL_FEATURES])
print(f'Processed Feature Matrix Shape: {X_trans.shape}')
```

### 3. Handle Class Imbalance with SMOTE


```python
y_diab = df['target_diabetes'].values
print(f'Before SMOTE: Positive={sum(y_diab==1)}, Negative={sum(y_diab==0)}')
X_bal, y_bal = balance_training_data(X_trans, y_diab)
print(f'After SMOTE: Positive={sum(y_bal==1)}, Negative={sum(y_bal==0)}')
```

### 4. Save DVC Versioned Dataset Manifest


```python
manifest = save_dataset_manifest(df)
print('Dataset Manifest SHA256:', manifest['sha256'])
```
