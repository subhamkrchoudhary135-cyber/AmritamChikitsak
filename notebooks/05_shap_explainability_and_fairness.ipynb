### 2. Run Inference & Extract Top 5 SHAP Factors


```python
test_input = ScreeningInput(
    age=48,
    gender='male',
    height_cm=168.0,
    weight_kg=78.0,
    systolic_bp=138.0,
    diastolic_bp=88.0,
    fasting_glucose=118.0,
    hba1c=6.1,
    physical_activity_hours_per_week=1.0,
    smoking_status='never',
    alcohol_consumption='none',
    family_history_diabetes=True,
)
conds, fore, pheno, shap_factors = model_registry.run_screening_inference(test_input)
print('Phenotype:', pheno.cluster_label)
print('Top 5 SHAP Contributing Factors:')
for f in shap_factors:
    print(f' - {f.feature_name_friendly}: {f.direction} (Impact: {f.impact_score})')
```

### 3. Fairlearn Fairness Evaluation


```python
for cond, fair in model_registry.fairness_reports.items():
    print(f'=== Fairness Report for {cond} ===')
    for attr, stats in fair.items():
        print(f'{attr}: Demographic Parity Diff = {stats.get("demographic_parity_difference")}, Passed = {stats.get("fairness_passed")}')
```
