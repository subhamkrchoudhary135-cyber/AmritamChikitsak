# 03 - 3-6 Month Trajectory Forecasting & Conformal Prediction
Forecasting future glycemic trajectories (Fasting Glucose mg/dL), blood pressure (Systolic BP mmHg), and continuous cardiometabolic risk with Ridge, Lasso, SVR, KNN, and XGBoost, with $(1 - \alpha) = 90\%$ distribution-free Conformal Prediction intervals.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
from src.ml.datasets import generate_synthetic_clinical_cohort
from src.ml.preprocessor import build_preprocessor_pipeline, NUMERIC_FEATURES, CATEGORICAL_FEATURES
from src.ml.regressors import ConformalBiomarkerForecaster, MultiTargetHealthForecaster
```

### 1. Fit Multi-Target Health Forecaster


```python
df = generate_synthetic_clinical_cohort(n_samples=1500, random_state=42)
preprocessor = build_preprocessor_pipeline()
X = preprocessor.fit_transform(df[NUMERIC_FEATURES + CATEGORICAL_FEATURES])
y_dict = {
    'target_glucose_6m': df['target_glucose_6m'].values,
    'target_sbp_6m': df['target_sbp_6m'].values,
    'target_continuous_risk': df['target_continuous_risk'].values,
}
forecaster = MultiTargetHealthForecaster(random_state=42)
forecaster.fit(X, y_dict)
print('Glucose Forecaster Metrics:', forecaster.glucose_forecaster.metrics)
```

### 2. Generate Prediction with Conformal Intervals


```python
sample_patient = X[:1]
pred = forecaster.forecast(sample_patient)
for k, v in pred.items():
    print(f'{k}: {v}')
```
